```python
import pandas as pd
import matplotlib as mp 

data = pd.read_csv('titanic.csv', sep=',')
df = pd.DataFrame(data)

# Preenchendo espaços vazios na idade com a média das idades
df['Age'].fillna(df['Age'].mean(), inplace= True)
```

```python
# 1) Distribuição de Idade por Gênero e Classe Social
# Agrupando a média, mediana e desvio padrão por idade e classe social 

df1_media = df.groupby('Age')['Pclass'].mean()
df1_mediana = df.groupby('Age')['Pclass'].median()
df1_desvio_padrao = df.groupby('Age')['Pclass'].std()

# Agrupando a média, mediana e desvio padrão por idade e sexo

# Substitui os valores de sexo por 0 (feminino) e 1 (masculino) para o cálculo da média
df.replace({'Sex': {'female': 0, 'male': 1}}, inplace=True)

df2_media = df.groupby('Age')['Sex'].mean()
df2_mediana = df.groupby('Age')['Sex'].median()
df2_desvio_padrao = df.groupby('Age')['Sex'].std()
```

```python
# 2) Taxa de Sobrevivência por Faixa Etária

# Classificando as idades em faixas etárias

df['Faixa_Etaria'] = pd.cut(df['Age'], bins=[0, 18, 30, 60, 120], labels=['Criança', 'Jovens', 'Adultos', 'Idosos'])

# Média dos sobreviventes por faixa etária
df3_media_faixa = df.groupby('Faixa_Etaria')['Survived'].mean()

# Número de sobreviventes por faixa etária
df3_numero_sobreviventes = df.groupby('Faixa_Etaria')['Survived'].sum()

# Taxa de sobrevivência por faixa etária
Taxa_sobrevivencia_criancas = (df3_numero_sobreviventes['Criança'] / df['Faixa_Etaria'].value_counts()['Criança']) * 100
print(f"Taxa de sobrevivência de crianças: {Taxa_sobrevivencia_criancas:.2f}%")
Taxa_sobrevivencia_jovens = (df3_numero_sobreviventes['Jovens'] / df['Faixa_Etaria'].value_counts()['Jovens']) * 100
print(f"Taxa de sobrevivência de jovens: {Taxa_sobrevivencia_jovens:.2f}%")
Taxa_sobrevivencia_adultos = (df3_numero_sobreviventes['Adultos'] / df['Faixa_Etaria'].value_counts()['Adultos']) * 100
print(f"Taxa de sobrevivência de adultos: {Taxa_sobrevivencia_adultos:.2f}%")
Taxa_sobrevivencia_idosos = (df3_numero_sobreviventes['Idosos'] / df['Faixa_Etaria'].value_counts()['Idosos']) * 100
print(f"Taxa de sobrevivência de idosos: {Taxa_sobrevivencia_idosos:.2f}%")
```

```python
# 3) Relação entre Tarifa e Sobrevivência

# Média de sobreviventes por faixa de tarifa
df4_media_tarifa = df.groupby('Fare')['Survived'].mean()
# Número de sobreviventes por faixa de tarifa
df['Faixa_Tarifa'] = pd.cut(df['Fare'], bins=[0, 20, 50, 100, 10000], labels=['0-20', '21-50', '51-100', '100+'])
df4_numero_sobreviventes_tarifa = df.groupby('Faixa_Tarifa')['Survived'].sum()

# Taxa de sobrevivência por faixa de tarifa
Taxa_sobrevivencia_barata = (df4_numero_sobreviventes_tarifa['0-20'] / df['Faixa_Tarifa'].value_counts()['0-20']) * 100
print(f"Taxa de sobrevivência para tarifas baratas: {Taxa_sobrevivencia_barata:.2f}%")
Taxa_sobrevivencia_intermediaria = (df4_numero_sobreviventes_tarifa['21-50'] / df['Faixa_Tarifa'].value_counts()['21-50']) * 100
print(f"Taxa de sobrevivência para tarifas intermediárias: {Taxa_sobrevivencia_intermediaria:.2f}%")
Taxa_sobrevivencia_executivo = (df4_numero_sobreviventes_tarifa['51-100'] / df['Faixa_Tarifa'].value_counts()['51-100']) * 100
print(f"Taxa de sobrevivência para tarifas executivas: {Taxa_sobrevivencia_executivo:.2f}%")
Taxa_sobrevivencia_primeira_classe = (df4_numero_sobreviventes_tarifa['100+'] / df['Faixa_Tarifa'].value_counts()['100+']) * 100
print(f"Taxa de sobrevivência para a primeira classe: {Taxa_sobrevivencia_primeira_classe:.2f}%")

# Valor médio da tarifa paga pelos sobreviventes
df4_valor_medio_tarifa = df[df['Survived'] == 1]['Fare'].mean()
print(f"Valor médio da tarifa paga pelos sobreviventes: {df4_valor_medio_tarifa:.2f}")

# Valor médio da tarifa paga pelos não sobreviventes
df4_valor_medio_tarifa_nao_sobreviventes = df[df['Survived'] == 0]['Fare'].mean()
print(f"Valor médio da tarifa paga pelos não sobreviventes: {df4_valor_medio_tarifa_nao_sobreviventes:.2f}")
```
