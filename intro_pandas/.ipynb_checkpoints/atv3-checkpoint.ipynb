```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('titanic.csv')
df.head()
```

```python
df.columns
```

```python
df['faixa_etaria'] = pd.cut(df['Age'], bins=[0, 18, 30, 60, 120], labels=['crianca', 'jovem', 'adulto', 'idoso'])
```

```python
df.head()
```

```python
plt.figure(figsize = (8, 6))
df[df['Sex'] == 'male']['Age'].hist(label = 'Male', grid = 'off', color = 'lightgreen')
df[df['Sex'] == 'female']['Age'].hist(label = 'Female', grid = 'off')
plt.grid('off', alpha = 0.7)
plt.legend()
plt.xlabel('Idade')
plt.ylabel('Num. de Pessoas')
plt.savefig('results/histograma_idade.png', dpi = 300)
plt.show()
```

```python
df['Age'].describe()
```

```python
distribuicao = df.groupby('Pclass')['Age'].describe()
```

```python
distribuicao
```

```python
df.boxplot(column='Age', by='Pclass')
plt.xlabel('Classe social')
plt.title(' ')
plt.subtitle(' ')
plt.ylabel('Idade')
plt.grid('off')
plt.savefig('results/boxplot_agrupado_classe_idade.png', dpi = 300)
plt.show()
```

```python
for faixa_i in df['faixa_etaria'].unique():
    taxa = df[df['faixa_etaria'] == faixa_i]['Survived'].sum()/df[df['faixa_etaria'] == faixa_i].shape[0]
    print(f"Taxa de sobrevivencia {faixa_i}: {taxa:.4f}")
```

```python
df_sobreviventes = df[df['Survived'] == 1]
df_sobreviventes
```

```python
df_mortos = df[df['Survived'] == 0]

plt.figure(figsize=(8,6))
df_mortos['Fare'].plot(kind = 'box', label = "Mortos")
plt.ylabel('Tarifa')
plt.savefig('results/boxplot_tarifa_mortos.png', dpi = 300)
plt.show()

plt.figure(figsize=(8,6))
df_sobreviventes['Fare'].plot(kind = 'box', label = "Sobreviventes")
plt.ylabel('Tarifa')
plt.savefig('results/boxplot_tarifa_sobreviventes.png', dpi = 300)
plt.show()
```

```python
plt.figure(figsize = (8,6))
plt.hist(df_mortos['Fare'], bins = 12, label = 'Mortos')
plt.hist(df_sobreviventes['Fare'], bins = 15, label = 'Sobreviventes', alpha = 0.75, color = 'lightgreen')
plt.legend()
plt.ylabel('Número de pessoas')
plt.xlabel('Valor da tarifa')
plt.savefig('results/histograma_tarifas.png', dpi = 300)
plt.show()
```
